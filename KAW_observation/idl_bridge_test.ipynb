```python
print("before import", flush=True)
from idlpy import *
print("after import", flush=True)

print("before IDL.run", flush=True)
out = IDL.run('print, "Hello from IDL via Python"')
print("after IDL.run", flush=True)
print("IDL output repr =", repr(out), flush=True)

out2 = IDL.run('print, !VERSION.RELEASE')
print("IDL version repr =", repr(out2), flush=True)
```

```python
IDL.run('x = findgen(10)')
IDL.run('y = x^2')

print("x =", IDL.x)
print("y =", IDL.y)
```

```python
import numpy as np

x = np.arange(10, dtype=np.float32)

IDL.x = x
IDL.run('y = sin(x)')

print("Python x =", x)
print("IDL y -> Python =", IDL.y)
```

```python
IDL.run("t0 = time_double('2022-09-01/00:00:00')")
```

```python
IDL.run("time = t0 + dindgen(100)*60d")
IDL.run("data = sin(findgen(100)*0.1)")
```
