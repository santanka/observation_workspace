# Araseの電場・磁場データのplot

```python
import pyspedas as psp
import pytplot as pt

pt.del_data('*')

time_range = ['20220901/21:00:00', '20220902/00:00:00']

psp.erg.pwe_efd(trange=time_range, level='l2', datatype='64', coord='dsi', get_support_data=True)
psp.erg.pwe_efd(trange=time_range, level='l2', datatype='64', coord='wpt', get_support_data=True)
psp.erg.pwe_efd(trange=time_range, level='l2', datatype='E_spin', get_support_data=True)

psp.erg.mgf(trange=time_range, level='l2', datatype='64hz', coord='dsi', get_support_data=True)
psp.erg.mgf(trange=time_range, level='l2', datatype='64hz', coord='sgi', get_support_data=True)
psp.erg.mgf(trange=time_range, level='l2', datatype='8sec', coord='dsi', get_support_data=True)
```

```python
time_range_T            = [time_range[0].replace('/', 'T'), time_range[1].replace('/', 'T')]

da_E64_dsi_Ex_waveform  = pt.data_quants['erg_pwe_efd_l2_E64Hz_dsi_Ex_waveform'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_E64_dsi_Ey_waveform  = pt.data_quants['erg_pwe_efd_l2_E64Hz_dsi_Ey_waveform'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))

da_E64_wpt_Eu_waveform  = pt.data_quants['erg_pwe_efd_l2_E64Hz_wpt_Eu_waveform'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_E64_wpt_Ev_waveform  = pt.data_quants['erg_pwe_efd_l2_E64Hz_wpt_Ev_waveform'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))

da_Espin_dsi_Eu_waveform    = pt.data_quants['erg_pwe_efd_l2_E_spin_Eu_dsi'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_Espin_dsi_Ev_waveform    = pt.data_quants['erg_pwe_efd_l2_E_spin_Ev_dsi'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))

da_B64_dsi_waveform     = pt.data_quants['erg_mgf_l2_mag_64hz_dsi'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_B64_sgi_waveform     = pt.data_quants['erg_mgf_l2_mag_64hz_sgi'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_Bspin_dsi_waveform   = pt.data_quants['erg_mgf_l2_mag_8sec_dsi'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
```

```python
da_E64_dsi_quality_flag = pt.data_quants['erg_pwe_efd_l2_E64Hz_dsi_quality_flag'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_E64_wpt_quality_flag = pt.data_quants['erg_pwe_efd_l2_E64Hz_wpt_quality_flag'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_Espin_dsi_quality_flag   = pt.data_quants['erg_pwe_efd_l2_E_spin_quality_flag'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))

da_B64_quality_flag     = pt.data_quants['erg_mgf_l2_quality_64hz'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
da_Bspin_quality_flag   = pt.data_quants['erg_mgf_l2_quality_8sec'].sortby('time').sel(time=slice(time_range_T[0], time_range_T[1]))
```

```python
import xarray as xr
import numpy as np

da_E64_dsi_Ex_waveform_qf   = da_E64_dsi_Ex_waveform.where(da_E64_dsi_quality_flag.interp(time=da_E64_dsi_Ex_waveform.time, method='nearest', kwargs={'fill_value': 'extrapolate'}) == 0, np.nan)
da_E64_dsi_Ey_waveform_qf   = da_E64_dsi_Ey_waveform.where(da_E64_dsi_quality_flag.interp(time=da_E64_dsi_Ey_waveform.time, method='nearest', kwargs={'fill_value': 'extrapolate'}) == 0, np.nan)

da_E64_wpt_Eu_waveform_qf   = da_E64_wpt_Eu_waveform.where(da_E64_wpt_quality_flag.interp(time=da_E64_wpt_Eu_waveform.time, method='nearest', kwargs={'fill_value': 'extrapolate'}) == 0, np.nan)
da_E64_wpt_Ev_waveform_qf   = da_E64_wpt_Ev_waveform.where(da_E64_wpt_quality_flag.interp(time=da_E64_wpt_Ev_waveform.time, method='nearest', kwargs={'fill_value': 'extrapolate'}) == 0, np.nan)

da_Espin_dsi_Eu_waveform_qf = da_Espin_dsi_Eu_waveform.where(da_Espin_dsi_quality_flag.interp(time=da_Espin_dsi_Eu_waveform.time, method='nearest', kwargs={'fill_value': 'extrapolate'}) == 0, np.nan)
da_Espin_dsi_Ev_waveform_qf = da_Espin_dsi_Ev_waveform.where(da_Espin_dsi_quality_flag.interp(time=da_Espin_dsi_Ev_waveform.time, method='nearest', kwargs={'fill_value': 'extrapolate'}) == 0, np.nan)

da_B64_dsi_waveform_qf      = xr.where(da_B64_quality_flag[:, 3] <= 21, da_B64_dsi_waveform, np.nan)
da_B64_sgi_waveform_qf      = xr.where(da_B64_quality_flag[:, 3] <= 21, da_B64_sgi_waveform, np.nan)
da_Bspin_dsi_waveform_qf    = xr.where(da_Bspin_quality_flag[:, 3] <= 21, da_Bspin_dsi_waveform, np.nan)
```

```python
import os
import numpy as np
from joblib import Parallel, delayed, cpu_count
import matplotlib as mpl
import matplotlib.pyplot as plt

mpl.use('Agg')

# 保存先
path_base_save_plot = '/mnt/j/KAW_observation/E_B_ratio_Arase/2022-09-01/2230-2330/waveform/4sec'
os.makedirs(path_base_save_plot, exist_ok=True)

# 走査パラメータ
t_scan_start = np.datetime64('2022-09-01T21:00:00')
t_scan_end   = np.datetime64('2022-09-01T23:59:00')   # 最終開始時刻の上限
step         = np.timedelta64(4, 's')                 # 8秒ステップ
win          = np.timedelta64(8, 's')                # 16秒窓

windows = [(t0, t0 + win) for t0 in np.arange(t_scan_start, t_scan_end + step, step)]

mpl.rcParams['font.size'] = 15

def plot_and_save(t0, t1, outdir):
    # --- 時間切り出し ---
    da_E64_dsi_Ex_waveform_qf_analysis = da_E64_dsi_Ex_waveform_qf.sel(time=slice(t0, t1))
    da_E64_dsi_Ey_waveform_qf_analysis = da_E64_dsi_Ey_waveform_qf.sel(time=slice(t0, t1))
    da_E64_wpt_Eu_waveform_qf_analysis = da_E64_wpt_Eu_waveform_qf.sel(time=slice(t0, t1))
    da_E64_wpt_Ev_waveform_qf_analysis = da_E64_wpt_Ev_waveform_qf.sel(time=slice(t0, t1))
    da_B64_dsi_waveform_qf_analysis    = da_B64_dsi_waveform_qf.sel(time=slice(t0, t1))
    da_B64_sgi_waveform_qf_analysis    = da_B64_sgi_waveform_qf.sel(time=slice(t0, t1))

    # データ無しならスキップ
    if da_E64_dsi_Ex_waveform_qf_analysis.sizes.get('time', 0) == 0:
        return

    # --- Figure ---
    fig = plt.figure(figsize=(10, 25))
    gs  = fig.add_gridspec(7, 1, hspace=0.15)
    ax_0 = fig.add_subplot(gs[0, 0])
    ax_1 = fig.add_subplot(gs[1, 0], sharex=ax_0)
    ax_2 = fig.add_subplot(gs[2, 0], sharex=ax_0)
    ax_3 = fig.add_subplot(gs[3, 0], sharex=ax_0)
    ax_4 = fig.add_subplot(gs[4, 0], sharex=ax_0)
    ax_5 = fig.add_subplot(gs[5, 0], sharex=ax_0)
    ax_6 = fig.add_subplot(gs[6, 0], sharex=ax_0)

    ax_0.plot(da_E64_dsi_Ex_waveform_qf_analysis.time, da_E64_dsi_Ex_waveform_qf_analysis.values, lw=1, c='red', label='x')
    ax_0.plot(da_E64_dsi_Ey_waveform_qf_analysis.time,  da_E64_dsi_Ey_waveform_qf_analysis.values, lw=1, c='blue', label='y')
    ax_1.plot(da_B64_dsi_waveform_qf_analysis.time,     da_B64_dsi_waveform_qf_analysis.values[:, 0], lw=1, c='red')
    ax_2.plot(da_B64_dsi_waveform_qf_analysis.time,     da_B64_dsi_waveform_qf_analysis.values[:, 1], lw=1, c='blue')
    ax_3.plot(da_B64_dsi_waveform_qf_analysis.time,     da_B64_dsi_waveform_qf_analysis.values[:, 2], lw=1, c='k')
    ax_4.plot(da_E64_wpt_Eu_waveform_qf_analysis.time,  da_E64_wpt_Eu_waveform_qf_analysis.values,     lw=1, c='orange', label='U')
    ax_4.plot(da_E64_wpt_Ev_waveform_qf_analysis.time,  da_E64_wpt_Ev_waveform_qf_analysis.values,     lw=1, c='green', label='V')
    ax_5.plot(da_B64_sgi_waveform_qf_analysis.time,     da_B64_sgi_waveform_qf_analysis.values[:, 0],  lw=1, c='red', label='x')
    ax_5.plot(da_B64_sgi_waveform_qf_analysis.time,     da_B64_sgi_waveform_qf_analysis.values[:, 1],  lw=1, c='blue', label='y')
    ax_6.plot(da_B64_sgi_waveform_qf_analysis.time,     da_B64_sgi_waveform_qf_analysis.values[:, 2],  lw=1, c='k')

    ax_0.set_ylabel('DSI 64 Hz\n[mV/m]')
    ax_1.set_ylabel('DSI 64 Hz Bx\n[nT]')
    ax_2.set_ylabel('DSI 64 Hz By\n[nT]')
    ax_3.set_ylabel('DSI 64 Hz Bz\n[nT]')
    ax_4.set_ylabel('WPT 64 Hz\n[mV/m]')
    ax_5.set_ylabel('SGI 64 Hz Bxy\n[nT]')
    ax_6.set_ylabel('SGI 64 Hz Bz\n[nT]')

    ax_0.legend()
    ax_4.legend()
    ax_5.legend()

    axes = [ax_0, ax_1, ax_2, ax_3, ax_4, ax_5, ax_6]
    for ax in axes:
        ax.minorticks_on()
        ax.grid(which='both', alpha=0.5, lw=1, linestyle=':')
    for ax in axes[:-1]:
        ax.tick_params(axis='x', which='both', labelbottom=False)

    axes[-1].set_xlim(t0, t1)

    fig.tight_layout()

    # --- 保存 ---
    t0s = np.datetime_as_string(t0, unit='s')
    t1s = np.datetime_as_string(t1, unit='s')
    fn  = f"waveform_{t0s.replace(':','')}_{t1s.replace(':','')}.png"
    fig.savefig(os.path.join(outdir, fn), dpi=300, bbox_inches='tight')
    plt.close(fig)

# ループ実行
n_jobs = max(1, cpu_count())
Parallel(n_jobs=n_jobs, backend="loky", verbose=5)(
    delayed(plot_and_save)(t0, t1, path_base_save_plot) for (t0, t1) in windows
)
```
