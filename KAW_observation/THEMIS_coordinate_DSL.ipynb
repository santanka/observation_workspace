# THEMIS-AのDSL座標とFAC座標系の関係を理解

# FAC座標系の定義
- z軸は、背景磁場$B_{0}$の単位ベクトルで与える。
- x軸は、反地球方向かつz軸と垂直な単位ベクトルで与える。
- y軸は、z軸とx軸の外積で与える。

# 反地球方向単位ベクトル (GSE座標系, DSL座標系) $\hat{\mathbf{r}}_{\mathrm{GSE/DSL}}$の決定

````python
import pyspedas as psp
import pytplot as pt
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import xarray as xr

pt.del_data('*')

trange = ['2022-09-01/21:00', '2022-09-01/23:59']

psp.themis.state(probe='a', trange=trange, no_update=True)

da_THA_pos_gse  = pt.data_quants['tha_pos_gse']
da_THA_pos_gse  = da_THA_pos_gse.sortby('time').sel(time=slice(trange[0], trange[1]))

print(da_THA_pos_gse)
````

````python
da_THA_pos_unit_gse = da_THA_pos_gse / np.sqrt((da_THA_pos_gse * da_THA_pos_gse).sum(dim='v_dim'))
print(da_THA_pos_unit_gse)
````

````python
pt.store_data('tha_pos_unit_gse', data={'x': da_THA_pos_unit_gse.time, 'y': da_THA_pos_unit_gse.data}, attr_dict=da_THA_pos_unit_gse.attrs)
psp.projects.themis.dsl2gse(name_in='tha_pos_unit_gse', name_out='tha_pos_unit_dsl', isgsetodsl=True, probe='a')

da_THA_pos_unit_dsl = pt.data_quants['tha_pos_unit_dsl']

print(da_THA_pos_unit_dsl)

print(np.sqrt((da_THA_pos_unit_dsl * da_THA_pos_unit_dsl).sum(dim='v_dim')))
````

# 背景磁場単位ベクトル (DSL座標系) $\hat{\mathbf{b}}_{0, \mathrm{DSL}}$ の決定

````python
import pyspedas as psp
import pytplot as pt
import xarray as xr
import numpy as np

psp.themis.fgm(trange=trange, probe='a', level='l2', no_update=True)

da_B_16_dsl = pt.data_quants['tha_fgl_dsl'].sortby('time').sel(time=slice(trange[0], trange[1]))
````

````python
background_time_sec = 100 #[sec]
````

````python
time_width_B_16Hz       = (da_B_16_dsl.time[10] - da_B_16_dsl.time[9]) / np.timedelta64(1, 's')
da_B_background         = da_B_16_dsl.rolling(time=int(background_time_sec/time_width_B_16Hz), center=True).mean('time')
da_B_background_unit    = da_B_background / np.sqrt((da_B_background * da_B_background).sum(dim='v_dim'))
da_B_background_unit    = da_B_background_unit.dropna(how='all', dim='time')

print(da_B_background_unit)
print(np.sqrt((da_B_background_unit * da_B_background_unit).sum(dim='v_dim')))
````

# DSL座標系で、FACの単位ベクトルを定義
```math
\mathbf{u} = \hat{\mathbf{r}}_{\mathrm{DSL}} - (\hat{\mathbf{r}}_{\mathrm{DSL}} \cdot \hat{\mathbf{b}}_{0, \mathrm{DSL}}) \hat{\mathbf{b}}_{0, \mathrm{DSL}}    \\
\mathbf{e}_{z, \mathrm{FAC}}    := \hat{\mathbf{b}}_{0, \mathrm{DSL}}   \\
\mathbf{e}_{x, \mathrm{FAC}}    := \frac{\mathbf{u}}{|\mathbf{u}|}      \\
\mathbf{e}_{y, \mathrm{FAC}}    := \mathbf{e}_{z, \mathrm{DSL}} \times \mathbf{e}_{x, \mathrm{DSL}}
```

````python
time_array  = da_B_background_unit.time

da_THA_pos_unit_dsl_interp    = da_THA_pos_unit_dsl.interp(time=time_array)

da_u_   = da_THA_pos_unit_dsl_interp - (da_THA_pos_unit_dsl_interp * da_B_background_unit).sum(dim='v_dim') * da_B_background_unit

da_e_z_FAC_inDSL    = da_B_background_unit.drop_attrs()
da_e_x_FAC_inDSL    = (da_u_ / np.sqrt((da_u_ * da_u_).sum(dim='v_dim'))).drop_attrs()
da_e_y_FAC_inDSL    = (xr.apply_ufunc(np.cross, da_e_z_FAC_inDSL, da_e_x_FAC_inDSL, input_core_dims=[['v_dim'], ['v_dim']], output_core_dims=[['v_dim']], vectorize=True)).drop_attrs()

print(da_e_x_FAC_inDSL)
print('')
print(da_e_y_FAC_inDSL)
print('')
print(da_e_z_FAC_inDSL)
````

# FAC座標系にDSL単位ベクトルを変換

````python
import xarray as xr
import numpy as np

# FAC の各軸を DSL で表したものを 3×3 行列にまとめる
R_FAC_to_DSL = xr.concat(
    [da_e_x_FAC_inDSL, da_e_y_FAC_inDSL, da_e_z_FAC_inDSL],
    dim='axis'
)
R_FAC_to_DSL    = R_FAC_to_DSL.assign_coords(axis=['x_fac', 'y_fac', 'z_fac'])
R_FAC_to_DSL    = R_FAC_to_DSL.dropna(dim='time', how='any')
# dims: (time, axis, v_dim)  = (time, FAC軸, DSI成分)

# DSL→FAC の回転行列（直交行列なので転置）
R_DSL_to_FAC    = R_FAC_to_DSL.transpose('time', 'v_dim', 'axis')

print(R_DSL_to_FAC)
````

````python
da_e_x_DSL_inFAC    = R_DSL_to_FAC.sel(v_dim=0)
da_e_y_DSL_inFAC    = R_DSL_to_FAC.sel(v_dim=1)
da_e_z_DSL_inFAC    = R_DSL_to_FAC.sel(v_dim=2)

print(da_e_z_DSL_inFAC)
````

````python
cos_theta_z = da_e_z_DSL_inFAC.isel(axis=2)
theta_z_deg = np.rad2deg(np.arccos(cos_theta_z))

import matplotlib as mpl
mpl.rcParams['font.size'] = 15

import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))
theta_z_deg.plot()
plt.minorticks_on()
plt.ylabel('angle between DSL-Z and FAC-Z [deg]')
plt.grid(True, which='both', linestyle=':')
````

````python
import matplotlib.pyplot as plt

it = 100000
ex = da_e_x_DSL_inFAC.isel(time=it)  # (axis: x_fac,y_fac,z_fac)
ey = da_e_y_DSL_inFAC.isel(time=it)
ez = da_e_z_DSL_inFAC.isel(time=it)

# FAC 成分を取り出しやすくしておく
ex_x = ex.sel(axis='x_fac').item()
ex_y = ex.sel(axis='y_fac').item()
ex_z = ex.sel(axis='z_fac').item()

ey_x = ey.sel(axis='x_fac').item()
ey_y = ey.sel(axis='y_fac').item()
ey_z = ey.sel(axis='z_fac').item()

ez_x = ez.sel(axis='x_fac').item()
ez_y = ez.sel(axis='y_fac').item()
ez_z = ez.sel(axis='z_fac').item()

fig, axs = plt.subplots(1, 3, figsize=(15, 5))  # 6x5 を 3 枚分

def setup_ax(ax, xlabel, ylabel, title):
    ax.axhline(0, color='k', linewidth=0.5)
    ax.axvline(0, color='k', linewidth=0.5)
    ax.set_aspect('equal', adjustable='box')
    ax.set_xlim(-1, 1)
    ax.set_ylim(-1, 1)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.minorticks_on()
    ax.grid(True, which='both', linestyle=':')
    ax.set_title(title)

# ---------------- (x, y) plane ----------------
ax = axs[0]
# FAC 基底
ax.quiver(0, 0, 1, 0, angles='xy', scale_units='xy', scale=1, color='k',   label='FAC-x')
ax.quiver(0, 0, 0, 1, angles='xy', scale_units='xy', scale=1, color='gray', label='FAC-y')
ax.quiver(0, 0, 0, 0, angles='xy', scale_units='xy', scale=1, color='purple', label='FAC-z', linewidth=0)
# DSI 軸
ax.quiver(0, 0, ex_x, ex_y, angles='xy', scale_units='xy', scale=1, color='r', label='DSL-x')
ax.quiver(0, 0, ey_x, ey_y, angles='xy', scale_units='xy', scale=1, color='b', label='DSL-y')
ax.quiver(0, 0, ez_x, ez_y, angles='xy', scale_units='xy', scale=1, color='g', label='DSL-z')

setup_ax(ax, 'FAC-x (Radial)', 'FAC-y (Longitudinal)', '(x, y) plane')
ax.legend(loc='upper left')

# ---------------- (x, z) plane ----------------
ax = axs[1]
ax.quiver(0, 0, 1, 0, angles='xy', scale_units='xy', scale=1, color='k',   label='FAC-x')
ax.quiver(0, 0, 0, 1, angles='xy', scale_units='xy', scale=1, color='purple', label='FAC-z')

ax.quiver(0, 0, ex_x, ex_z, angles='xy', scale_units='xy', scale=1, color='r', label='DSL-x')
ax.quiver(0, 0, ey_x, ey_z, angles='xy', scale_units='xy', scale=1, color='b', label='DSL-y')
ax.quiver(0, 0, ez_x, ez_z, angles='xy', scale_units='xy', scale=1, color='g', label='DSL-z')

setup_ax(ax, 'FAC-x (Radial)', 'FAC-z (Parallel)', '(x, z) plane')

# ---------------- (y, z) plane ----------------
ax = axs[2]
ax.quiver(0, 0, 1, 0, angles='xy', scale_units='xy', scale=1, color='gray',   label='FAC-y')
ax.quiver(0, 0, 0, 1, angles='xy', scale_units='xy', scale=1, color='purple', label='FAC-z')

ax.quiver(0, 0, ex_y, ex_z, angles='xy', scale_units='xy', scale=1, color='r', label='DSL-x')
ax.quiver(0, 0, ey_y, ey_z, angles='xy', scale_units='xy', scale=1, color='b', label='DSL-y')
ax.quiver(0, 0, ez_y, ez_z, angles='xy', scale_units='xy', scale=1, color='g', label='DSL-z')

setup_ax(ax, 'FAC-y (Longitudinal)', 'FAC-z (Parallel)', '(y, z) plane')

fig.suptitle(str(da_e_x_DSL_inFAC.time.values[it]), fontsize=14)
plt.tight_layout()
plt.show()
````
